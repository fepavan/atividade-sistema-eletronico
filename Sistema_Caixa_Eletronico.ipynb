````python
# Sistema de Caixa Eletrônico

def depositar(saldo):
    valor = float(input("Informe o valor do depósito: "))
    if valor > 0:
        saldo += valor
        print(f"Depósito realizado! Novo saldo: R$ {saldo:.2f}")
    else:
        print("Valor inválido para depósito.")
    return saldo

def sacar(saldo):
    valor = float(input("Informe o valor do saque: "))
    if valor > saldo:
        print("Saldo insuficiente para este saque!")
    elif valor <= 0:
        print("Valor inválido para saque.")
    else:
        saldo -= valor
        print(f"Saque realizado! Novo saldo: R$ {saldo:.2f}")
    return saldo

# Programa Principal
print("\n==== Bem-vindo ao GEN-CAIXA ====")
saldo = float(input("Informe seu saldo inicial: "))

# O loop 'while True' faz com que o menu continue a aparecer após cada operação
while True:
    print("\n--- Menu ---")
    print("1 - Ver saldo")
    print("2 - Depositar")
    print("3 - Sacar")
    print("4 - Sair")

    pergunta = int(input("Qual opção você quer usar? "))

    if pergunta == 1:
        print(f"Seu saldo atual é de: R$ {saldo:.2f}")
    elif pergunta == 2:
        saldo = depositar(saldo)  # Chama a função e atualiza o saldo
    elif pergunta == 3:
        saldo = sacar(saldo)      # Chama a função e atualiza o saldo
    elif pergunta == 4:
        print("Obrigado por usar o GEN-CAIXA. Até logo!")
        break                     # Sai do loop e encerra o programa
    else:
        print("Opção inválida. Tente novamente.")
````
